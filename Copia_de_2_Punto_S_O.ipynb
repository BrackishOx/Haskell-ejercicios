```python
import multiprocessing
import time
import random
from datetime import datetime

def imprimir_estado(msg):
    print(f"[{datetime.now().strftime('%H:%M:%S')}] {msg}", flush=True)

def esperar_bus(nombre, semaforo, flag_ocupado):
    while True:
        if semaforo.acquire(block=False):
            with flag_ocupado.get_lock():
                flag_ocupado.value = 0
            return
        else:
            with flag_ocupado.get_lock():
                if flag_ocupado.value == 0:
                    imprimir_estado(f"{nombre} detectó que el bus está ocupado.")
                    flag_ocupado.value = 1
        time.sleep(0.4)

def simular_dispositivo(nombre, semaforo, flag_ocupado):
    while True:
        esperar_bus(nombre, semaforo, flag_ocupado)

        try:
            duracion_envio = round(random.uniform(1.0, 3.5), 2)
            inicio = time.time()
            imprimir_estado(f"{nombre} está usando el bus... ")
            time.sleep(duracion_envio)
            duracion_real = time.time() - inicio
            imprimir_estado(f"{nombre} liberó el bus. Tiempo de uso: {duracion_real:.2f} segundos.\n")
        finally:
            semaforo.release()

        time.sleep(random.expovariate(1 / 2.5))

def main():
    bus = multiprocessing.Semaphore(1)
    flag_ocupado = multiprocessing.Value('i', 0)

    nombres = ["Nodo-1", "Nodo-2", "Nodo-3", "Nodo-4", "Nodo-5"]
    procesos = []

    for nombre in nombres:
        p = multiprocessing.Process(target=simular_dispositivo, args=(nombre, bus, flag_ocupado))
        p.start()
        procesos.append(p)

    try:
        for p in procesos:
            p.join()
    except KeyboardInterrupt:
        imprimir_estado("Interrupción detectada. Finalizando procesos...")

if __name__ == "__main__":
    main()
```
